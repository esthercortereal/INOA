# Nova seção Exercicío: Analise do desempenho de clubes no campeonato brasileiro

**Objetivo:** os alunos irão aprendera:


*   Manipular dados de um campeonato de futebol
*   Calcular e vizualizar estatisticas de desempenho dos clubes

# Nova seção Passo 1: Leitura da planilha

Carregar os dados do arquivo brasileirão.xlsx usando pandas




```python
import pandas as pd

df = pd.read_excel("brasileirao.xlsx")

df.head()
```

# Nova seção Passo 2: Calcular o saldo de gols (SG)

Adicionar uma nova coluna chamada "Saldo de Gols" que será a diferença entre Gols Pró(GP) e Gols Contra(GC).

```python
df['Saldo de Gols'] = df['GP'] - df['GC']
df.head()
```

# Nova seção Passo 3: Caslcular a pontuação total

Criar uma coluna chamada 'Pontos' onde a pontuação será calculada da seguinte forma:



* Vitória (V) = 3 pontos
* Empate (E) = 1 ponto
* Derrota (D) = 0 pontos



```python
df['Pontos'] = df['V'] * 3 + df['E']

df.head(20)
```

# Nova seção Filtra os clubes com mais de 40 pts

filtrar os clubes que tem mais de 40 pontos

```python
clubes_maior_40_pontos = df[df['Pontos'] > 40]

clubes_maior_40_pontos
```

#  Passo 5: vizualizar os dados

Criar um gráfico de barras paea visualizar a pontuação dos clubes


```python
df.plot(x='Clube', y='Pontos', kind='bar', title='Pontuação dos Clubes no Brasileirão 2024')
```
